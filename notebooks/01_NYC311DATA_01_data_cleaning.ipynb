```python
print(df.columns)
```

 Clean the text columns
Remove extra spaces at the start and end

```python
df["created_date"] = df["created_date"].str.strip()
df["agency"] = df["agency"].str.strip()
df["complaint_type"] = df["complaint_type"].str.strip()
df["descriptor"] = df["descriptor"].str.strip()
df["borough"] = df["borough"].str.strip()
df["status"] = df["status"].str.strip()
df["closed_date"] = df["closed_date"].str.strip()
```

3b. Make the category columns all capitals

```python
df["agency"] = df["agency"].str.upper()
df["borough"] = df["borough"].str.upper()
df["status"] = df["status"].str.upper()
```

3c. Check the categories

```python
df["borough"].value_counts()
```

```python
df["status"].value_counts()
```

Convert dates from text to real dates
errors="coerce" turns bad dates into empty values (NaT) instead of crashing.


```python
df["created_date"] = pd.to_datetime(df["created_date"], errors="coerce")
df["closed_date"] = pd.to_datetime(df["closed_date"], errors="coerce")
df.info()
```

Find problem rows
Closed before it was created (impossible)

```python
df["bad_dates"] = df["closed_date"] < df["created_date"]
df["bad_dates"].sum()
```

Location outside New York

```python
df["bad_location"] = (df["latitude"] < 40.4) | (df["latitude"] > 41.0) | \
                     (df["longitude"] < -74.3) | (df["longitude"] > -73.6)
df["bad_location"].sum()
```

Is the request closed?
notna(): Short for "not available" or "not a number". It checks each cell in the column and asks: "Is this cell filled with real data?"

```python
df["is_closed"] = df["closed_date"].notna()
df["is_closed"].value_counts()
```

How many hours did it take to close?
3,600 seconds in one hour (60 seconds × 60 minutes), dividing the total seconds by 3,600 converts the seconds into hours.

```python
df["resolution_hours"] = (df["closed_date"] - df["created_date"]).dt.total_seconds() / 3600
```

Remove the hours for bad-date rows (keep the rows)
#df.loc[..., ...]: This is a Pandas tool used to target specific rows and columns
#df["bad_dates"]: This is a list of True/False values
#"resolution_hours": Specifies that we only want to change the values in this particular column.
# = np.nan: Short for Not a Number. It replaces the bad numbers with a blank/missing value placeholder.

```python
df.loc[df["bad_dates"], "resolution_hours"] = np.nan
df["resolution_hours"].describe()
```

Time columns for trend analysis


```python
df["month"] = df["created_date"].dt.month
df["weekday"] = df["created_date"].dt.day_name()
df["hour"] = df["created_date"].dt.hour
df.head()
```

```python
print("From:", df["created_date"].min())
print("To:  ", df["created_date"].max())
```
